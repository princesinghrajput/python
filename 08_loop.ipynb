```python
# Running outer loop from 2 to 3
for i in range(2, 21):
 
    # Printing inside the outer loop
    # Running inner loop from 1 to 10
    for j in range(1, 11):
    
      # Printing inside the inner loop
      print(i, "*", j, "=", i*j)
    # Printing inside the outer loop
    print()
```
