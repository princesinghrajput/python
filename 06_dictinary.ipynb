## Nested dictionary

```python
myfamily = {
  "child1" : {
    "name" : "Emil",
    "year" : 2004
  },
  "child2" : {
    "name" : "Tobias",
    "year" : 2007
  },
  "child3" : {
    "name" : "Linus",
    "year" : 2011,
    "child":{
        "name":"child3 ka bachha",
        "age":21

    }
  }
}

myfamily.items()
```

```python
child1 = {
  "name" : "Emil",
  "year" : 2004
}



child2 = {
  "name" : "Tobias",
  "year" : 2007
}


child3 = {
  "name" : "Linus",
  "year" : 2011
}


myfamily = {
    "childA":child1,
    "childB":child2,
    "childC":child3
}

print(myfamily.items())
```
