```python
class Employee:
    def __init__(self, name, id):
        self.name = name
        self.id = id



class Develope(Employee):
    def __init__(self, name, id, language):
        super().__init__(name, id)
        self.language = language
      


prince = Employee("Prince kumar", "123")

prince_dev = Develope("mohan", "456", "Python")

print(prince.name)

print(prince_dev.name)
```
