## POLYMORPHISM
    >> DUCK TYPING
    - Python uses duck typing to check if an object has a method it needs.

    >> OPERATOR OVERLOADING
    - Python supports operator overloading. You can define your own way of handling operators like +, -, *, /, etc.

    >> METHOD OVERRIDING
    - Python supports method overriding. You can override a method in a subclass with a method with the same name in the superclass.

```python
class VsCode:
    def execute(self):
        print('compiling')
        print('running')


class Laptop:

    def code(self, ide):
        ide.execute()

class MyEditor:
    def execute(self):
        print('Executing spell check')
        print('Convention check')
        print('Formatting code')
        print('Linting code')

ide = MyEditor() # ide = VsCode()

lap1 = Laptop()

lap1.code(ide)
```
