```python
import random  # Import the random module to generate random numbers

def guessing_game():
    number_to_guess = random.randint(1, 100)  # Generate a random number between 1 and 100
    attempts = 0  # Initialize the number of attempts
    
    print("Welcome to the Guessing Game!")
    print("I'm thinking of a number between 1 and 100.")
    
    while True:  # Start an infinite loop until the correct guess is made
        guess = input("Please enter your guess: ")  # Get user input
        
        if not guess.isdigit():  # Check if the input is not a digit
            print("That's not a valid number. Please try again.")  # Provide feedback for invalid input
            continue  # Skip the rest of the loop and ask for input again
        
        guess = int(guess)  # Convert the valid input to an integer
        attempts += 1  # Increment the attempts count
        
        if guess < number_to_guess:  # Check if the guess is lower than the number
            print("Too low! Try again.")  # Provide feedback
        elif guess > number_to_guess:  # Check if the guess is higher than the number
            print("Too high! Try again.")  # Provide feedback
        else:  # The guess is correct
            print(f"Congratulations! You've guessed the number {number_to_guess} in {attempts} attempts.")
            break  # Exit the loop when the guess is correct

# Call the function to start the game
guessing_game()
```
