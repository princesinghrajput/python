## Create a function to take three number as input and return the sum of the number

```python

def sum_of_three():
    a = float(input("Enter first number"))

    b = int(input("Enter second number"))

    c = float(input("Enter third number"))

    total=a+b+c

    return total

print(sum_of_three())




```
