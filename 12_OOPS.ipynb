```python
class Computer:

    def __init__(self, cpu, ram):

        self.cpu= cpu
        self.ram=ram
        

    def config(self):
        print("Config is: ", self.cpu, self.ram)

comp1 = Computer("Ryzen", 16)

comp1.config()

comp2 = Computer("i5", 8)

comp2.config()

```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age


    def about(self):
        print(f"{self.name} is {self.age} years old")


person1 = Person("Prince", 21)

person1.age = 40

person1.about()

print(person1)




```

```python
class A():
    def feature1(self):
        print("Feature 1 is working")


class B(A):
    def feature2(self):
        print("Feature 2 is working")


class C(A,B):
    def feature3(self):
        print("Feature 3 is working")


b1= B()
c1=C()

c1.feature3()

b1.feature1()

```

```python
class Animal:
    def speak(self):
        print("Animal speaking..")


class Dog(Animal):
    def bark(self):
        print("Dog barking..")

dog1=Dog()
dog1.speak()
dog1.bark()
```

## Multi level Inheritance

```python
class Animal:
    def speak(self):
        print("Animal speaking..")


class Dog(Animal):
    def bark(self):
        print("Dog barking..")

class DogChild(Dog):
    def play(self):
        print("Dog playing..")

d1=DogChild()

d1.speak()
d1.play()
d1.bark()
```

## Multiple Inheritance

```python
class A:
    def feature1(self):
        print("Feature 1 is working")

class B:
    def feature2(self):
        print("Feature 2 is working")

class C:
    def feature3(self):
        print("Feature 3 is working")

class D(A,B,C):
    def feature4(self):
        print("Feature 4 is working")

d1 = D()

d1.feature4()
```
